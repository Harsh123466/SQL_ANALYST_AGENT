```python
import duckdb
import pandas as pd

# %pip install duckdb
```

```python
# Create DuckDB connection
db = duckdb.connect("sales.duckdb")

# load dataset
retail_sales = pd.read_csv(
    "../data/Red30 Tech US online retail sales.csv"
)

regional_sales = pd.read_csv(
    "../data/Red30 Tech US sales by region.csv"
)
```

```python
# Register DataFrames in DuckDB
db.register("retail_sales", retail_sales)
db.register("regional_sales", regional_sales)
```

```python
print(db.execute("SELECT COUNT(*) FROM retail_sales").fetchone())
print(db.execute("SELECT COUNT(*) FROM regional_sales").fetchone())
```

```python
print("Retail Sales:", retail_sales.shape)
print("Regional Sales:", regional_sales.shape)
```

```python
print(db.sql("DESCRIBE retail_sales").fetchdf())
```

```python
print(db.sql("DESCRIBE regional_sales").fetchdf())
```
